```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
```

```python
data=pd.read_csv('data/stud.csv')
print(data.head())
```

```python
data.isna().sum()
```

```python
data.describe()
```

```python
data['total_score'] = data['math score'] + data['reading score'] + data['writing score']
data['average_score'] = data['total_score'] / 3
```

```python
data['math_smarties'] = data['math score'] > data['math score'].mean()
```

```python
fig,axs=plt.subplots(1,2,figsize=(12,5))
plt.subplot(1,2,1)
sns.boxplot(data=data, x='average_score', ax=axs[0])
plt.subplot(1,2,2)
sns.histplot(data=data, x='average_score', bins=20, ax=axs[1])
plt.show()  
```
